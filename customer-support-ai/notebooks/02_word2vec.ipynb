# Train Skip-Gram Word2Vec Model

This notebook trains a custom Skip-Gram Word2Vec model on the preprocessed customer support ticket dataset.

```python
import pandas as pd
import os
import multiprocessing
from gensim.models import Word2Vec

# 1. Load Preprocessed Data
data_path = "../artifacts/data_transformation/transformed_dataset.csv"
df = pd.read_csv(data_path)

# 2. Prepare tokens for Word2Vec
# Word2Vec expects a list of lists of tokens
sentences = [str(text).split() for text in df['body'].fillna("")]

print(f"Total sentences to train on: {len(sentences)}")
print(f"Sample sentence: {sentences[0][:10]}")
```

```python
# 3. Train the Skip-Gram Word2Vec Model
print("Training Word2Vec model...")
cores = multiprocessing.cpu_count()

w2v_model = Word2Vec(
    sentences=sentences,
    vector_size=100,    # Embedding size: 100
    window=5,           # Window size: 5
    sg=1,               # sg=1 (Skip-Gram)
    min_count=2,        # min_count=2
    workers=cores       # Use all available CPU cores
)
print("Training completed.")
```

```python
# 4. Save the trained model
os.makedirs("../artifacts/word2vec", exist_ok=True)
model_path = "../artifacts/word2vec/skipgram_v100.model"
w2v_model.save(model_path)
print(f"Model saved successfully to {model_path}")
```

```python
# 5. Experiment: Test Embeddings
word = "invoice"
if word in w2v_model.wv:
    similar_words = w2v_model.wv.most_similar(word)
    print(f"Most similar words to '{word}':")
    for w, score in similar_words:
        print(f"  - {w}: {score:.4f}")
else:
    print(f"Word '{word}' not in vocabulary.")
```
